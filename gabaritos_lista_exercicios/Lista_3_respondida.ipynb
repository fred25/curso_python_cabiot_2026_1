```python
# Já vou importar as bibliotecas
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# e vou carregar o dataset
dados = sns.load_dataset("iris")
```

## Sugestões de análise!

- Vocês podem aproximar a área da sépala e da pétala multiplicando seu comprimento pela largura
- Vocês podem criar razões dividindo uma coluna pela outra
- Compare as distribuições!

> Lembrando que nessa lista não existe certo ou errado, o que importa é fazer uma análise válida que respondam as perguntas, vou fazer a análise proposta acima

```python
# Boa sorte nessa jornada!
# aproximar área
dados["sepal_area"] = dados["sepal_length"] * dados["sepal_width"]
dados["petal_area"] = dados["petal_length"] * dados["petal_width"]
dados.head()
```

```python
dados["razao_area"] = dados["sepal_area"] / dados["petal_area"]
dados.head()
```

```python
# podemos ver que a área das petalas separa bem as 3 espécies
sns.scatterplot(dados, x="sepal_area", y="petal_area", hue="species")
sns.kdeplot(dados, x="sepal_area", y="petal_area", hue="species", fill=True, alpha=0.3)
plt.show()
```

```python
# esse gráfico evidencia isso
sns.boxplot(dados, x="species", hue="species", y="petal_area")
plt.show()
```
