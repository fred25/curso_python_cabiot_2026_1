## Exercício 1

Escreva uma função que diz se um número é par ou não

```python
# Definindo funcao que ve se o numero é par
def eh_par(num : float) -> bool:
    """
    Função que verifica se um número é par
    """
    # Verifica se o resto da divisão por dois é 0
    if num % 2 == 0:
        # se for 0, é par
        return True
    # senão
    else:
        # não é par
        return False

# Mostrando resultado na tela
print(eh_par(2))
```

## Exercício 2

Um mercado organiza seu estoque em um dicionário do python, onde as chaves são os nomes dos produtos e os valores é a quantidade no estoque.

Escreva uma função que receba um dicionário com produtos e quantidades e o nome de um produto e mostre na tela:
- `"Tudo ok!"` se o produto tiver 10 ou mais em estoque
- `"Acabando"` se o produto tiver menos de 5 em estoque
- `"Indisponível"` se o produto tiver 0 no estoque
- `"Produdo inválido"` se o produto não estiver nas chaves do dicionário
- Se a quantidade for exatamente 10 e o produto for carne mostre: `"Prepara o churrasco"` e não nenhuma mensagem acima
- Se tiver entre 5 e 10 não mostre nada.

A função não deve retornar nada

```python
# Dicionário de estoque
estoque = {"Maçã" : 25, "Arroz" : 5, "Carne" : 10, "Milho" : 10, "Carvão" : 3}

# Defina a função do exercício
def verifica_estoque(estoque : dict, produto : str):
    """
    Função que verficia o estoque da loja
    """
    # Verifico que o produto existe na loja
    if produto in estoque.keys():
        # defino a quantidade do meu produto como uma variável
        quantidade = estoque[produto]

        if quantidade == 10 and produto == "Carne":
            print("Prepara o churrasco")
        elif quantidade >= 10:
            print("Tudo ok!")
        elif quantidade < 5:
            print("Acabando")
        elif quantidade == 0:
            print("Indisonível")
    
    # se não tiver o produto no dicionario
    else:
        print("Produto inválido")


# Teste valores na resposta
print(verifica_estoque(estoque, "Carne"))
```

## Exercício 3

Crie uma função que recebe uma lista somente de números e retorna a média dos valores dessa lista

```python
# Definindo uma lista teste
lista = [10, 20, 33, 56, 30, 100, 28, 99, 39]

# Definindo a função que calcula a média
def calcula_media(lista : list) -> float:
    """
    Função que calcula a média de valores em uma lista
    """
    # defino a soma como 0
    soma = 0
    # somo todos os valores da lista
    for item in lista:
        soma += item
    
    # divido a soma dos valores pelo tamaho da lista
    media = soma / len(lista)

    return media

# Definindo o resultado
media = calcula_media(lista)

# mostrando na tela
print(media)
```

## Exercício 4

Crie uma função que receba uma lista e devolva o menor valor dentro dela

```python
# Definindo lista de teste
lista = [0, -1, -67, 100, 29, 88, 2, 4, 19]

# Definindo função que acha o menor valor

def acha_menor_valor(lista : list):
    """
    Função que acha o menor valor em uma lista
    """
    # defino o valor atual como infinito
    menor = float("+inf")

    # comparo o valor atual com todos os elementos da lista
    for elemento in lista:
        if elemento < menor:
            # se o elemento for menor que o menor, menor vira meu elemento
            menor = elemento
    
    # retorno o menor elemento
    return menor

# Definindo o menor valor
menor_valor = acha_menor_valor(lista)

# mostrando na tela
print(menor_valor)
```
