## Funções

Funções são blocos de código reutilizáveis que executam uma tarefa específica.

### Estrutura de uma função

```python
def nome_da_funcao(parametro_1: tipo_1, parametro_2: tipo_2) -> tipo_retorno:
    """
    Documentação da função.
    Pode ter várias linhas.
    """

    # Corpo da função

    return valor
```

### Exemplos de funções

- `print()` → mostra uma mensagem na tela  
- `type()` → retorna o tipo de um valor

```python
print("Mensagem")
type(17)
```

### Chamando uma função

Para usar uma função, escreva seu nome seguido dos argumentos entre parênteses.

```python
print("Olá!") -> Não retorna nada, mas mostra olá na tela
type(3.14) -> Retorna float
nome_da_funcao(17, "Mensagem") -> retorna algo do tipo tipo_retorno
```

### Salvando o retorno de uma função

Funções que possuem `return` podem ter seus resultados armazenados em variáveis.

```python
tipo_numero = type(17)

print(tipo_numero)
```
`int` 

Nesse exemplo, o valor retornado por `type(17)` é salvo na variável `tipo_numero`.
