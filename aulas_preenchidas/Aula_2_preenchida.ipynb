## Fluxo não linear: Condições

Podemos fazer escolhas no código e para isso faremos a seguinte estrutura:
```python
if condição:
    bloco caso condição seja verdadeira
elif outra condição
    bloco caso outra condição seja verdadeira
else
    bloco caso tudo seja falso
```
---
Vamos implementar a arvore de decisão do slide:

````python
# Vamos definir as características da festa
festa_boa = False
comida_gratis = True

# implemente aqui a arvore de decisão
# Sea festa for boa
if festa_boa:
    # Ficamos na festa
    print("Ficar na festa")
# Se não for boa, mas a comida é grátis
elif comida_gratis:
    # Ficamos também!
    print("Ficar na festa")
# Senão ...
else:
    # Vamos embora
    print("Ir embora")
````

As condições também funcionam com operações lógicas:

````python
# Verifique se as idades são iguais
minha_idade = 20
tua_idade = 25

# Implemente aqui a verificação se temos idades iguais
if minha_idade == tua_idade:
    print("Nossa! temos a mesma idade")
else:
    print("Eh, não temos a mesma idade")
````

## Controle de fluxo - loops

Como podemos fazer coisas repetidas vezes?

Utilizando loops, as suas principais estruturas são:
- `while`
- `for`

---

Exemplo de estrutura `while`:
```python
while condição:
    bloco de código
```
Executa o bloco de código enquanto a condição for verdadeira

````python
# Crie um pequeno contador com while:
# crio uma variavel para manter registro de qual numero eu contei
contador = 0
# crio um loop que roda até o contador chegar em 10
while contador < 10:
    # mostro o valor do contador
    print(contador)
    # adiciono um ao contador
    contador += 1
````

Exemplo de estrutura `for`:
```python
for valor in estrutura:
    bloco de código
```
Executa o bloco de código uma vez para cada componente em uma estrutura de dados.

````python
def c_to_k(t_celsius : float) -> float:
    return t_celsius + 273

# Com for aplique e mostre na tela o resultado da função c_to_k em todas as temperaturas da lista
temperaturas = [10, 15, 15.4, -18, 34, 19.2]

# faca um for que printe o resultado em cada um dos componentes da lista
for temp in temperaturas:
    print(c_to_k(temp))
````

## Desafio

Vocês devem criar uma função que receba uma sequência de RNA, considere que ela começa com AUG
Ela deve mostrar o resultado da proteína codificada por esse RNA.

Considere que a tabela dos códons está organizada no dicionario `tabela_codons`

````python
# Definindo a tablea dos códons
tabela_codons = {'UUU': 'F', 'CUU': 'L', 'AUU': 'I', 'GUU': 'V',
                 'UUC': 'F', 'CUC': 'L', 'AUC': 'I', 'GUC': 'V',
                 'UUA': 'L', 'CUA': 'L', 'AUA': 'I', 'GUA': 'V',
                 'UUG': 'L', 'CUG': 'L', 'AUG': 'M', 'GUG': 'V',
                 'UCU': 'S', 'CCU': 'P', 'ACU': 'T', 'GCU': 'A',
                 'UCC': 'S', 'CCC': 'P', 'ACC': 'T', 'GCC': 'A',
                 'UCA': 'S', 'CCA': 'P', 'ACA': 'T', 'GCA': 'A',
                 'UCG': 'S', 'CCG': 'P', 'ACG': 'T', 'GCG': 'A',
                 'UAU': 'Y', 'CAU': 'H', 'AAU': 'N', 'GAU': 'D',
                 'UAC': 'Y', 'CAC': 'H', 'AAC': 'N', 'GAC': 'D',
                 'UAA': 'Stop', 'CAA': 'Q', 'AAA': 'K', 'GAA': 'E',
                 'UAG': 'Stop', 'CAG': 'Q', 'AAG': 'K', 'GAG': 'E',
                 'UGU': 'C', 'CGU': 'R', 'AGU': 'S', 'GGU': 'G',
                 'UGC': 'C', 'CGC': 'R', 'AGC': 'S', 'GGC': 'G', 
                 'UGA': 'Stop', 'CGA': 'R', 'AGA': 'R', 'GGA': 'G',
                 'UGG': 'W', 'CGG': 'R', 'AGG': 'R', 'GGG': 'G'}

# Definindo a sequência teste
sequencia = "AUGGAAUCUACUCGUGAAUAA"

# Definindo a função
def traduz_proteina(seq : str) -> str:
    """
    Função que dada uma sequência de RNA, retorna a proteína codificada por essa sequência
    """
    # crio uma proteina vazia
    proteina = ""
    # crio uma variavel para marcar a minha posição
    posicao = 0
    # crio um loop que roda até a minha posição ser o final da sequencia
    while posicao < len(seq):
        #defino o codon que estou vendo da minha posição até ela +3
        codon = seq[posicao : posicao + 3]
        # defino o meu aminoácido olhando na tabela
        aminoacido = tabela_codons[codon]
        # verifico se o aminoácido é 'Stop"
        if aminoacido == "Stop":
            # se for, paro o loop
            break
        # senão
        else:
            # adiciono o aminoácido na proteína
            proteina += aminoacido
        
        # antes de terminar o loop ando 3 casas pra frente
        posicao += 3
    
    # ao final do loop retorno a proteina
    return proteina

# Mostrando o resultado na tela
print(traduz_proteina(sequencia))
````
