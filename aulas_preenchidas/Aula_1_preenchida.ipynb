## Funções

Funções são blocos de código reutilizáveis que executam uma tarefa específica.

### Estrutura de uma função

```python
def nome_da_funcao(parametro_1: tipo_1, parametro_2: tipo_2) -> tipo_retorno:
    """
    Documentação da função.
    Pode ter várias linhas.
    """

    # Corpo da função

    return valor
```

### Exemplos de funções

- `print()` → mostra uma mensagem na tela  
- `type()` → retorna o tipo de um valor

```python
print("Mensagem")
type(17)
```

### Chamando uma função

Para usar uma função, escreva seu nome seguido dos argumentos entre parênteses.

```python
print("Olá!") -> Não retorna nada, mas mostra olá na tela
type(3.14) -> Retorna float
nome_da_funcao(17, "Mensagem") -> retorna algo do tipo tipo_retorno
```

### Salvando o retorno de uma função

Funções que possuem `return` podem ter seus resultados armazenados em variáveis.

```python
tipo_numero = type(17)

print(tipo_numero)
```
`int` 

Nesse exemplo, o valor retornado por `type(17)` é salvo na variável `tipo_numero`.

### Próximo exercício

Crie uma função que receba uma temperatura em graus Celsius como argumento e retorne o valor convertido em Kelvin.

````python
# Definindo uma temperatura para teste
t_c = 25

# Defina a função que converte celsius para Kelvin
def c_to_k(temp_celsius : float) -> float:
    """
    Função que converte uma temperatura em celsius 
    para uma temperatura em kelvin
    """

    # converte celsius para kelvin
    temp_kelvin = temp_celsius + 273

    return temp_kelvin

# Salve o resultado em uma variável
t_k = c_to_k(t_c)

# Mostre o resultado na tela
print(t_k)
````

## Desafio 2

Faça uma função que converta celsius para fahrenheit.

> Curiosidade!

$-40 ºC$ é a mesma temperatura que $-40 ºF$ e é a única temperatura em que isso acontece.

````python
# Definindo uma temperatura para teste
t_c = -40

# Defina a função que converte celsius para Fahrenheit
def c_to_f(temp_celsius : float) -> float:
    """
    Função que converte uma temperatura em celsius 
    para uma temperatura em fahrenheit
    """

    # converte celsius para fahrenheit
    temp_fahrenheit = 9/5 * temp_celsius + 32

    return temp_fahrenheit

# Salve o resultado em uma variável
t_f = c_to_f(t_c)

# Mostre o resultado na tela
print(t_f)
````

## Resumo!

Vamos revisar tudo o que aprendemos na aula criando uma função que calcule o conteúdo GC de uma sequência de DNA.

> Curiosidade

As ligações entre as bases G com C no DNA são mais fortes que as de A com T, por isso utilizamos o conteúdo GC (porcentagem de C e G na sequência) como uma medida de estabilidade desse fragmento.

````python
seq = "AGGACACGACTCGCATAGGACTCCGAATAGGCTAGCTAGGACTCTCTGA"

# Vamos escrever nossa função:
def conteudo_GC(sequencia : str) -> float:
    """
    Função que recebe uma sequência de DNA e retorna
    seu conteúdo GC
    """

    # calcula o total de letras na sequência
    total = len(sequencia)

    # Calcula quantos C e G tem
    CG_contagem = sequencia.count("C") + sequencia.count("G")

    # calcula a proporção de CG
    proporcao = CG_contagem / total

    return proporcao

# Mostrar o valor na tela
print(conteudo_GC(seq))
````
